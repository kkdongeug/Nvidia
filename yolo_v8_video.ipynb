```python
# 2. 유튜브 영상 다운로드
youtube_url = "https://www.youtube.com/watch?v=RkDZJ1qQaxw"

# 영상 파일명을 video.mp4로 저장
!yt-dlp -f best -o "video.mp4" "{youtube_url}"
```

```python
# 3. YOLOv8으로 객체 감지
from ultralytics import YOLO

# YOLO 모델 불러오기 (필요에 따라 yolov8s.pt, yolov8m.pt 등 사용 가능)
model = YOLO('yolov8n.pt')

# 영상 감지 실행
results = model.predict(
    source='video.mp4',
    save=True,
    conf=0.3
)
```
